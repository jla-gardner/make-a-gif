```python
import matplotlib.pyplot as plt
import numpy as np

from make_a_gif import gif


def circle(t: float) -> None:
    x, y = np.cos(t), np.sin(t)
    plt.figure(figsize=(1, 1))
    plt.clf()
    plt.plot(x, y, "ok")
    plt.xlim(-1.3, 1.3)
    plt.ylim(-1.3, 1.3)
    plt.xticks([])
    plt.yticks([])


gif(
    np.linspace(0, 2 * np.pi, 100),
    circle,
    fps=30,
    savefig_kwargs={"transparent": True, "dpi": 600},
    css={"width": "100px"},
)
```
